```python
def multiply(a: int, b: int) -> int:
    """Multiply a and b.

    Args:
        a: first int
        b: second int
    """
    return a * b

llm_with_tools = llm.bind_tools([multiply])
```

```python
tool_call = llm_with_tools.invoke([HumanMessage(content=f"What is 7 multiplied by 30", name="Ian")])
```
